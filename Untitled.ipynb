```python
import tensorflow as tf
from tensorflow.keras import layers
import tensorflow_model_optimization as tfmot

IMG_HEIGHT = 224 
IMG_WIDTH = 224  
NUM_CLASSES = 6  

inputs = layers.Input(shape=(IMG_HEIGHT, IMG_WIDTH, 1))

x = layers.Conv2D(3, (1, 1), padding='same')(inputs)
x = layers.Rescaling(1./127.5, offset=-1)(x)

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMG_HEIGHT, IMG_WIDTH, 3),
    include_top=False,
    weights='imagenet'
)

base_model.trainable = False
x = base_model(x, training=False)

x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.2)(x)
outputs = layers.Dense(NUM_CLASSES, activation='softmax')(x)

float_model = tf.keras.Model(inputs, outputs)

WEIGHTS_PATH = 'steel_defect_clf_mobile_netv2_fine_tuning_scaling_patch_compat.weights.h5'
float_model.load_weights(WEIGHTS_PATH)

float_model.summary()
```

```python
quantize_model = tfmot.quantization.keras.quantize_model
qat_model = quantize_model(float_model)
```
